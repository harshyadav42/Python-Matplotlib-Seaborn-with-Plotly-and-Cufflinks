```python
import plotly.express as px
import seaborn as sns

tips = sns.load_dataset('tips')

fig = px.line(
    tips,
    y='total_bill'
)

fig.show()
```

```python
tips.groupby('day')['tip'].mean().iplot(kind = 'bar')
```

```python
data = tips.groupby('day', observed=True)['tip'].mean().reset_index()

fig = px.bar(data, x='day', y='tip', title='Average Tip by Day')
fig.show()
```

```python
fig = px.scatter(tips,x='total_bill', y='tip',title='Total Bill vs Tip')
fig.show()
```

```python

fig = px.box(tips, x='total_bill',y='day',title='Total Bill Distribution by Day')
fig.show()
```
